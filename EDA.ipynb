```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
zero_columns = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for column in zero_columns:
    print(column, ":", (df[column] == 0).sum())
```

```python
sns.countplot(x='Outcome', data=df)

plt.title("Distribution of Diabetes Outcome")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Outcome', y='Glucose', data=df)

plt.title("Glucose Level vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Glucose Level")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Outcome', y='BMI', data=df)

plt.title("BMI vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("BMI")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')

plt.title("Correlation Heatmap of PIMA Diabetes Dataset")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x='Glucose',
    y='BMI',
    hue='Outcome',
    data=df
)

plt.title("Glucose vs BMI Based on Diabetes Outcome")

plt.show()
```

```python
outcome_analysis = df.groupby('Outcome').mean()

print(outcome_analysis)
```

```python
outcome_analysis.T.plot(kind='bar', figsize=(12,6))

plt.title("Average Feature Values by Diabetes Outcome")
plt.xlabel("Features")
plt.ylabel("Average Value")

plt.xticks(rotation=45)

plt.show()
```
